```python
from satellite_images_nso.functions import functions 
import glob
import rasterio
import numpy as np
```

```python
for file in glob.glob("E:/data/zuidland/*.tif"):
    print("-------")
    print(file)
    functions.plot_tif(file)
    
    with rasterio.open(file) as src:
        # Read the image data
        image = src.read()
        
    data = (image[2] - image[2].min()) / (image[2].max() - image[2].min())
    print("High pixels: ")
    print(np.count_nonzero(data > np.max(data)*0.5))
    
    print("Mean pixel:")
    print(np.mean(data))
    print("Std pixel:")
    print(np.std(data))
```
