# Cloud checks.

This notebook checks if there are clouds in satellite images.
By plotting them and using a model on them.

```python
import glob
from cloud_recognition.api import detect_clouds
import pickle
from satellite_images_nso.other import functions
import cloud_checks_settings
import os
```

```python
model_download_url = "https://e34a505986aa74678a5a0e0f.blob.core.windows.net/satellite-images-nso/Cloud_detection_models/cloud_detection_logistic_regression_v1.0.sav"
cloud_model_path = os.path.abspath(os.path.join(os.getcwd(), '..')).replace("\\","/")+"/models/"+"cloud_detection_logistic_regression_v1.0.sav"
directory_regex= cloud_checks_settings.directory_regex
```

```python
if not os.path.exists(cloud_model_path):
    functions.download_file(model_download_url,cloud_model_path)
else:
    print("Model has already been downloaded")
```

```python
path_to_model = pickle.load(
        open(
            cloud_model_path,
            "rb",
        )
    )
```

```python
no_clouds = []
for afile in glob.glob(directory_regex ):
    afile = afile.replace("\\","/")
    print("---------")
    print(afile)
    try:
        if detect_clouds(model=path_to_model, filepath=afile) is True:
            no_clouds.append(afile)
    except Exception as e:
        print(e)
```

```python
no_clouds
```

```python
for file in no_clouds:
    file = file.replace("\\","/")
    print("--------------")
    print(file)
    functions.plot_tif(file)
```

```python


with open('no_clouds_schippers_gat.pkl', 'wb') as f:
    pickle.dump(no_clouds, f)
```
